Backpack with capacity 10kg.

Items:
 - laptop - 3kg,  8€
 - camera - 4kg, 10€
 - tent   - 5kg, 12€
 - jacket - 2kg,  5€
 - food   - 3kg,  7€

```python
from qiskit_optimization.applications import Knapsack
```

```python
# Define the problem
capacity = 10
values = [8, 10, 12, 5, 7]
weights = [3, 4, 5, 2, 3]

knapsack = Knapsack(
    values=values,
    weights=weights,
    max_weight=capacity
)

problem = knapsack.to_quadratic_program()

print(problem.prettyprint())
```
