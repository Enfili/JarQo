```python
import pennylane as qp
```

```python
def deutsch_oracle(case: int):
    if case not in [1, 2, 3, 4]:
        raise ValueError("`case` must be 1, 2, 3, or 4.")

    if case in [2, 3]:
        qp.CNOT(wires=[0, 1])
    
    if case in [3, 4]:
        qp.PauliX(wires=1)

dev = qp.device("default.qubit", wires=2)

@qp.qnode(dev)
def deutsch_circuit(case: int):
    qp.Hadamard(wires=0)
    qp.PauliX(wires=1)
    qp.Hadamard(wires=1)

    deutsch_oracle(case)

    qp.Hadamard(wires=0)
    return qp.probs(wires=0)

def deutsch_algorithm(case: int):
    probabilities = deutsch_circuit(case)
    return "Constant function" if probabilities[0] > 0.5 else "Balanced function"
```

```python
for case in range(1, 5):
    result = deutsch_algorithm(case)
    print(f"Case {case}: {result}")
```
